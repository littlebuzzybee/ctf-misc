```python
times = []
for ts in re.findall("\[(.*) \+0200\]", s):
    times.append(pd.to_datetime(ts, format="%d/%b/%Y:%H:%M:%S"))
```

```python
def long_sleep_2_bits(long_sleep):
    if long_sleep == 0:
        return "00"
    elif long_sleep == 2:
        return "01"
    elif long_sleep == 4:
        return "10"
    elif long_sleep == 6:
        return "11"
    else:
        print(f"long_sleep must be 0, 2, 4 or 6, was {long_sleep}")
        return ""
    
def short_sleep_2_bits(short_sleep):
    if short_sleep == 2:
        return "0"
    elif short_sleep == 4:
        return "1"
    else:
        print(f"short_sleep must be 2 or 4, was {short_sleep}")
        return ""
```

```python
password = ""
for i in range(21):
    print(i, end=" ")
    c = ""
    c += long_sleep_2_bits(sleeps[4*i])
    c += long_sleep_2_bits(sleeps[4*i+1])
    c += long_sleep_2_bits(sleeps[4*i+2])
    c += short_sleep_2_bits(sleeps[4*i+3])
    password += chr(int(c,2))
print(password)
```
